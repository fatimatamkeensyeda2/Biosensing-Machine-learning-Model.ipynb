```python
features = ["Protein_19",  "Protein_20", "Metabolite_20", "Gene_14", "Target_Analyte_Concentration_ng_mL"]
data = df[features].dropna()
corr = data.corr(method="spearman")
plt.figure(figsize=(10, 7))
sns.heatmap( corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, center=0,  fmt=".3f",  linewidths=0.5,  square=True)
plt.title(  "Correlation Heatmap: Candidate Multi-Omics Features and Target Analyte", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
# Save figure for GitHub
plt.savefig("biosensor_correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# The above heatmap indicate moderate positive association between candidate molecular features and target analyte.
```

# Candidate Features Analysis

```python
candidate_features = [ "Gene_14", "Protein_19", "Protein_20", "Metabolite_14", "Metabolite_18", "Metabolite_20"]
target = "Target_Analyte_Concentration_ng_mL"
for feature in candidate_features:
 sns.lmplot(data=df, x=feature, y=target, height=5, aspect=1.2 )
plt.title(f"{feature} vs Target Analyte Concentration")
plt.tight_layout()
plt.show()
```

# Spearman correlation coefficients and false discovery rate (FDR) correction:

```python
from scipy.stats import spearmanr
from statsmodels.stats.multitest import multipletests
results = []
for feature in candidate_features: r, p = spearmanr(df[feature], df["Target_Analyte_Concentration_ng_mL"], nan_policy="omit")
results.append([feature, r, p])
candidate_results = pd.DataFrame( results, columns=["Feature", "Spearman_r", "P_value"])
candidate_results["FDR"] = multipletests(
candidate_results["P_value"], method="fdr_bh")[1]
candidate_results = candidate_results.sort_values( "Spearman_r", ascending=False)
display(candidate_results)
```

```python
#Among the tested candidate features, Metabolite_20 showed a statistically significant moderate positive association with the target analyte after FDR correction.
```
